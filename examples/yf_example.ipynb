# Yahoo Finance quote example

This notebook requests quotes for a list of symbols using the `StockMonitor` package. It makes live network requests to Yahoo Finance.

```python
from StockMonitor.providers import YFinanceProvider
```

```python
symbols = [
    "AAPL",
    "MSFT",
    "GOOGL",
    "AMZN",
    "NVDA",
    "META",
    "TSLA",
    "BRK-B",
    "JPM",
    "V",
    "XOM",
    "WMT",
    "COST",
    "MA",
    "HD",
    "PG",
    "JNJ",
    "ABBV",
    "ORCL",
    "NFLX",
]
```

```python
provider = YFinanceProvider()
quotes = provider.get_quotes(symbols)

for quote in quotes:
    if quote is not None:
        print(quote["symbol"], quote["price"])
```
